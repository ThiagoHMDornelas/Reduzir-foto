```python
from PIL import Image
import os

# Pasta com as imagens
input_folder = input("Digite o diretório de origem") #"C:/Users/Thiago/Downloads/@Torrent/jobs/paraiso_conjuntos"
output_folder = input("Digite o diretório de saida") #"C:/Users/Thiago/Downloads/@Torrent/fotos reduzidas"

if (input_folder != '') and (output_folder != ''):
    # Criar a pasta de saída, se não existir
    os.makedirs(output_folder, exist_ok=True)

    # Reduzir cada imagem
    for filename in os.listdir(input_folder):
        if filename.lower().endswith(".jpg") or filename.lower().endswith(".jpeg"):
            img = Image.open(os.path.join(input_folder, filename))
            img = img.convert("RGB")  # Garante compatibilidade
            img.save(os.path.join(output_folder, filename), "JPEG", quality=70)  # Ajuste a qualidade

    print("Imagens reduzidas com sucesso!")
```
